```python
import json
import pandas as pd



actor_name_map = {}
actor_genre_map = {}


with open("OneDrive - University of Maryland/Fall 2026/INST414/Lab 2/imdb_movies_2000to2022.prolific.json", "r") as in_file:
    for line in in_file:
        this_movie = json.loads(line)

        for actor_id,actor_name in this_movie['actors']:
            actor_name_map[actor_id] = actor_name

            this_actors_genre = actor_genre_map.get(actor_id, {})
            for genre in this_movie["genres"]:
              this_actors_genre[genre] = this_actors_genre.get(genre, 0) + 1
            actor_genre_map[actor_id] = this_actors_genre

actor_genre_map["nm1165110"]



index = actor_genre_map.keys()
data = [actor_genre_map[k] for k in index]
df = pd.DataFrame(data, index=index)
df = df.fillna(0)
df.fillna(0, inplace=True)
df



query_id = "nm1165110"
query_vector = df.loc[[query_id]]

from sklearn.metrics import DistanceMetric
dist = DistanceMetric.get_metric('euclidean')

distances = dist.pairwise(query_vector, df)[0]
distance_series = pd.Series(distances, index = df.index)
top_10_ids = distance_series.drop(index=query_id).nsmallest(10)

for actor_id in top_10_ids.index:
  print(f"{actor_name_map[actor_id]} ({actor_id}): {top_10_ids[actor_id]:.4f}")
```
