```python
import json
import pandas as pd
from sklearn.cluster import KMeans
```

```python
actor_name_map = {}

with open("imdb_movies_2000to2022.prolific.json", "r") as in_file:
  for line in in_file:
    this_movie = json.loads(line)
    for actor_id, actor_name in this_movie["actors"]:
      actor_name_map[actor_id] = actor_name

df = pd.read_csv("imdb_movies_2000to2022.actorXgenre.csv", index_col = "actor_id")

cluster_model = KMeans(n_clusters=8, random_state=42)
cluster_labels = cluster_model.fit_predict(df)

actor_cluster_df = pd.DataFrame(cluster_labels, index=df.index, columns=["cluster"])

for cluster, actors in actor_cluster_df.groupby("cluster"):
  print(f"cluster: {cluster}, size: {len(actors)}")
  for actor_id in actors.sample(5).index:
    print("\t", actor_id, actor_name_map[actor_id])
```
