```python
%matplotlib inline

import pandas as pd
import json

actor_to_genre_df = pd.read_csv("imdb_movies_2000to2022.actorXgenre.csv", index_col="actor_id")

actor_to_genre_df
```

```python
from sklearn.cluster import KMeans

k = 8 # We pick k=8 for illustrative purposes only.
```

```python
inertias = []

for k in range(2, 31):
    model = KMeans(n_clusters=k, random_state=42, n_init=1)
    model.fit(actor_to_genre_df)
    inertias.append(model.inertia_)
```

```python
import matplotlib.pyplot as plt

plt.plot(range(2, 31), inertias, marker="o")
plt.xlabel("Number of Clusters (k)")
plt.ylabel("Inertia")
plt.title("Elbow Method")
plt.show()
```
